```python
nums = [4,3,2,7,8,2,3,1]

def element_twice(nums):
  output=[]
  for i in range(len(nums)):
    for j in range(i):
      if nums[i]==nums[j]:
        output.append(nums[j])
  return output
```

Question 2: Write a Python function that implements Merge Sort and provide an example demonstrating how it works.

```python
numbers = [7,2,4,5,7,2,8,9]

def merge_sort(numbers):
    if len(numbers) <= 1:
        return numbers

    middle = len(numbers) // 2
    left_half = numbers[:middle]
    right_half = numbers[middle:]

    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)

    return merge(left_sorted, right_sorted)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

```python
merge_sort(numbers)
```
